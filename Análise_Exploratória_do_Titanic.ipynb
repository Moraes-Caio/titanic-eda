```python
# Remoção de Outlier

# Feminino
sexfem_aiq = df_sexfem.fare.quantile(0.75) - df_sexfem.fare.quantile(0.25)

sexfem_ls = df_sexfem.fare.quantile(0.75) + 1.5 * sexfem_aiq
sexfem_li = df_sexfem.fare.quantile(0.25) - 1.5 * sexfem_aiq

df_sexfem_fc = df_sexfem[(df_sexfem.fare < sexfem_ls) & (df_sexfem.fare > sexfem_li)]
display(df_sexfem_fc.fare.describe())

cv_sexfem = (df_sexfem_fc.fare.std() / df_sexfem_fc.fare.mean()) * 100
print('\nCV do valor pago nas tarifas pelo sexo feminino: ', round(cv_sexfem,2), '%')

# Masculino
sexmale_aiq = df_sexmale.fare.quantile(0.75) - df_sexmale.fare.quantile(0.25)

sexmale_ls = df_sexmale.fare.quantile(0.75) + 1.5 * sexmale_aiq
sexmale_li = df_sexmale.fare.quantile(0.25) - 1.5 * sexmale_aiq

df_sexmale_fc = df_sexmale[(df_sexmale.fare < sexmale_ls) & (df_sexmale.fare > sexmale_li)]
display(df_sexmale_fc.fare.describe())

cv_sexmale = (df_sexmale_fc.fare.std() / df_sexmale_fc.fare.mean()) * 100
print('\nCV do valor pago nas tarifas pelo sexo masculino: ', round(cv_sexmale,2), '%')
```
